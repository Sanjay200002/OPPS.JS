    area() {

// subclass for ractangular

class Rectangle extends Shape {
    constructor(width, height) {
        super("Rectangle");
        this.width = width;
        this.height = height;
    }

    area() {
        return this.width * this.height;
    }

    perimeter() {
        return 2 * (this.width * this.height);
    }
}

// subclass for trangle

class RightTriangle extends Shape {
    constructor(width, height) {
        super("RightTriangle");
        this.width = width;
        this.height = height;
    }

    area() {
        return 0.5 * this.width * this.height;
    }

    perimeter() {
        const hypotenious = Math.sqrt (this.width **2 + this.height **2);
        return this.width + this.height + hypotenious;
